```python
calendar_sql = """select
    date,
    wm_yr_wk,
    weekday,
    wday,
    month,
    year,
    event_name_1,
    event_type_1,
    event_name_2,
    event_type_2,
    snap_CA,
    snap_TX,
    snap_WI
from `phonic-chemist-509603-v9.m5_raw.calendar`
"""

with open("models/staging/stg_calendar.sql", "w") as f:
    f.write(calendar_sql)

print("stg_calendar.sql written")
```

```python
prices_sql = """select
    store_id,
    item_id,
    wm_yr_wk,
    sell_price
from `phonic-chemist-509603-v9.m5_raw.sell_prices`
where sell_price > 0
"""

with open("models/staging/stg_sell_prices.sql", "w") as f:
    f.write(prices_sql)

print("stg_sell_prices.sql written")
```

```python
day_cols = ", ".join([f"d_{i}" for i in range(1, 1914)])

sales_sql = f"""select
    id,
    item_id,
    dept_id,
    cat_id,
    store_id,
    state_id,
    day_column,
    cast(replace(day_column, 'd_', '') as int64) as day_num,
    units_sold
from `phonic-chemist-509603-v9.m5_raw.sales_train_validation`
unpivot (
    units_sold for day_column in ({day_cols})
)
"""

with open("models/staging/stg_sales_unpivoted.sql", "w") as f:
    f.write(sales_sql)

print("stg_sales_unpivoted.sql written, length:", len(sales_sql))
```

```python
daily_mart_sql = """select
    s.id,
    s.item_id,
    s.dept_id,
    s.cat_id,
    s.store_id,
    s.state_id,
    c.date,
    c.wday,
    c.weekday,
    c.month,
    c.year,
    c.event_name_1,
    c.event_type_1,
    c.snap_CA,
    c.snap_TX,
    c.snap_WI,
    p.sell_price,
    s.units_sold,
    s.units_sold * p.sell_price as revenue
from {{ ref('stg_sales_unpivoted') }} s
left join {{ ref('stg_calendar') }} c
    on s.day_num = row_number() over (order by c.date)
left join {{ ref('stg_sell_prices') }} p
    on s.store_id = p.store_id
    and s.item_id = p.item_id
    and c.wm_yr_wk = p.wm_yr_wk
"""

with open("models/marts/mart_daily_sales.sql", "w") as f:
    f.write(daily_mart_sql)

print("mart_daily_sales.sql written")
```

```python
daily_mart_sql = """select
    s.id,
    s.item_id,
    s.dept_id,
    s.cat_id,
    s.store_id,
    s.state_id,
    c.date,
    c.wday,
    c.weekday,
    c.month,
    c.year,
    c.event_name_1,
    c.event_type_1,
    c.snap_CA,
    c.snap_TX,
    c.snap_WI,
    p.sell_price,
    s.units_sold,
    s.units_sold * p.sell_price as revenue
from {{ ref('stg_sales_unpivoted') }} s
inner join {{ ref('stg_calendar') }} c
    on s.day_num = c.day_num
left join {{ ref('stg_sell_prices') }} p
    on s.store_id = p.store_id
    and s.item_id = p.item_id
    and c.wm_yr_wk = p.wm_yr_wk
"""

with open("models/marts/mart_daily_sales.sql", "w") as f:
    f.write(daily_mart_sql)

print("mart_daily_sales.sql written")
```

```python
weekly_mart_sql = """select
    store_id,
    cat_id,
    dept_id,
    wm_yr_wk,
    min(date) as week_start,
    sum(units_sold) as total_units_sold,
    sum(revenue) as total_revenue,
    avg(sell_price) as avg_sell_price
from {{ ref('mart_daily_sales') }}
group by store_id, cat_id, dept_id, wm_yr_wk
"""

with open("models/marts/mart_weekly_sales.sql", "w") as f:
    f.write(weekly_mart_sql)

print("mart_weekly_sales.sql written")
```
