```python
calendar_sql = """select
    date,
    wm_yr_wk,
    weekday,
    wday,
    month,
    year,
    event_name_1,
    event_type_1,
    event_name_2,
    event_type_2,
    snap_CA,
    snap_TX,
    snap_WI
from `phonic-chemist-509603-v9.m5_raw.calendar`
"""

with open("models/staging/stg_calendar.sql", "w") as f:
    f.write(calendar_sql)

print("stg_calendar.sql written")
```

```python
prices_sql = """select
    store_id,
    item_id,
    wm_yr_wk,
    sell_price
from `phonic-chemist-509603-v9.m5_raw.sell_prices`
where sell_price > 0
"""

with open("models/staging/stg_sell_prices.sql", "w") as f:
    f.write(prices_sql)

print("stg_sell_prices.sql written")
```

```python
day_cols = ", ".join([f"d_{i}" for i in range(1, 1914)])

sales_sql = f"""select
    id,
    item_id,
    dept_id,
    cat_id,
    store_id,
    state_id,
    day_column,
    cast(replace(day_column, 'd_', '') as int64) as day_num,
    units_sold
from `phonic-chemist-509603-v9.m5_raw.sales_train_validation`
unpivot (
    units_sold for day_column in ({day_cols})
)
"""

with open("models/staging/stg_sales_unpivoted.sql", "w") as f:
    f.write(sales_sql)

print("stg_sales_unpivoted.sql written, length:", len(sales_sql))
```
